# Descubriendo Pi a través del Azar

Para entender cómo funciona la simulación de Montecarlo que calcula Pi ($\pi$), no necesitamos matemáticas avanzadas; solo necesitamos recordar un poco de geometría básica y usar la lógica de la probabilidad.

Este *notebook* es interactivo. Primero repasaremos la teoría y luego, en la parte inferior, ejecutaremos el código para simular y graficar el experimento.

### Paso 1: El escenario geométrico

Tenemos dos figuras geométricas:
1. **Un círculo**
2. **Un cuadrado** que encierra perfectamente a ese círculo.

Vamos a asignarle una medida. Digamos que el radio del círculo es $r$.
Como el cuadrado encierra perfectamente al círculo, el lado del cuadrado mide exactamente el diámetro del círculo, es decir, $2r$.

* **Área del círculo:** $A_c = \pi \cdot r^2$
* **Área del cuadrado:** $A_s = \text{lado} \times \text{lado} = (2r) \cdot (2r) = 4r^2$

### Paso 2: La proporción mágica

¿Qué pasa si dividimos el área del círculo entre el área del cuadrado para ver qué proporción del cuadrado está ocupada por el círculo?

$$ \text{Proporción} = \frac{\text{Área del círculo}}{\text{Área del cuadrado}} = \frac{\pi \cdot r^2}{4r^2} $$

Como tenemos $r^2$ arriba y abajo, podemos cancelarlos:

$$ \text{Proporción} = \frac{\pi}{4} $$

Esto significa que el círculo ocupa exactamente **$\pi / 4$** del área total del cuadrado.

### Paso 3: Entra Montecarlo (La Probabilidad)

Dejamos la geometría estática y entramos al mundo del azar.

Si lanzamos miles de puntos de forma completamente aleatoria y uniforme dentro del cuadrado, la probabilidad de que un punto caiga dentro del círculo es exactamente igual a la proporción de área que acabamos de calcular.

$$ \frac{\text{Puntos dentro del círculo}}{\text{Total de puntos lanzados}} \approx \frac{\pi}{4} $$

Si despejamos $\pi$ (pasando el 4 a multiplicar), obtenemos nuestra fórmula:

$$ \pi \approx 4 \cdot \left( \frac{\text{Puntos dentro del círculo}}{\text{Total de puntos lanzados}} \right) $$

### Paso 4: ¿Cómo sabe el computador si el punto cayó dentro?

Usamos un plano cartesiano y el famoso **Teorema de Pitágoras**.

Imagina que el centro es la coordenada $(0,0)$ y el radio es $1$. Cuando el computador genera un punto al azar $(x, y)$, calculamos su distancia al centro usando la fórmula: 
$$ d = \sqrt{x^2 + y^2} $$

Si $d \leq 1$, ¡el dardo cayó dentro del círculo!

## 💻 Simulación Interactiva en Python

Ejecuta las siguientes dos celdas de código. La primera importará las librerías necesarias y la segunda generará la gráfica interactiva.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider
```

```python
def simular_pi(n_puntos):
    # 1. Generar coordenadas x e y aleatorias entre -1 y 1
    x = np.random.uniform(-1, 1, n_puntos)
    y = np.random.uniform(-1, 1, n_puntos)
    
    # 2. Calcular la distancia al origen usando Pitágoras
    distancia = np.sqrt(x**2 + y**2)
    
    # 3. Identificar puntos dentro (distancia <= 1)
    dentro = distancia <= 1
    fuera = distancia > 1
    
    # 4. Calcular la estimación de Pi
    pi_estimado = 4 * np.sum(dentro) / n_puntos
    error = abs(np.pi - pi_estimado)
    
    # 5. Configurar el lienzo para la gráfica
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Dibujar los puntos (azules adentro, rojos afuera)
    ax.scatter(x[dentro], y[dentro], color='blue', s=3, alpha=0.5, label='Dentro del círculo')
    ax.scatter(x[fuera], y[fuera], color='red', s=3, alpha=0.5, label='Fuera del círculo')
    
    # Dibujar el contorno del círculo
    circulo = plt.Circle((0, 0), 1, color='black', fill=False, linewidth=2)
    ax.add_patch(circulo)
    
    # Ajustes visuales de los ejes
    ax.set_xlim(-1.05, 1.05)
    ax.set_ylim(-1.05, 1.05)
    ax.set_aspect('equal')
    
    # Título dinámico
    titulo = f'Puntos: {n_puntos} | Pi Estimado: {pi_estimado:.5f} | Error: {error:.5f}'
    ax.set_title(titulo, fontsize=14, fontweight='bold', pad=15)
    ax.legend(loc='upper right')
    
    # Mostrar gráfica
    plt.show()

# 6. Crear el control interactivo (slider)
interact(simular_pi, n_puntos=IntSlider(min=100, max=50000, step=500, value=1000, description='Puntos:'));
```
